```python
import pandas as pd
from pathlib import Path
import os
```

```python
df = pd.read_csv(root / "csv/ejercicios_clasificados.csv")
```

```python
varios_tipos_ejercicios_index = df[df.tipo_ejercicio.str.contains("|", regex=False, na = False)].index
```

```python
df.tipo_ejercicio = df.tipo_ejercicio.str.split("|")
```

```python
df = df.explode("tipo_ejercicio")
```

```python
df.tema.unique()
```

```python
def add_tipo_ejercicio(row):
    if pd.notna(row.tipo_ejercicio):
        year_pos = row.id.find("2")
        new_id = row.id[:year_pos] + row.tipo_ejercicio.replace(" ", "_") + "_" + row.id[year_pos:]
        return new_id
    else:
        return row.id
```

```python
new_id = df.apply(add_tipo_ejercicio, axis = 1)
df.id = new_id
```
