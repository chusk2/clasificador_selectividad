```python
def check_cinetica(statement):
    if not pd.isna(statement):
        for word in ["v = k", "velocidad de reacción", "catalizador",
        "energía de activación", "constante de velocidad", "cinética", "orden total",
    "ecuación de velocidad"]:
            if word in statement:
                return "cinética"
    return "equilibrio químico"
```

```python
df_cinetica["tema"] = df_cinetica.enunciado.apply(check_cinetica)
```

```python
acido_base_dict = {"disoluciones" : ["riqueza", "densidad", "disolución comercial", "industrial", "molaridad" ],
"neutralización" : ["neutraliza"],
"verdadero o falso" : ["verdadero", "veracidad", "falsedad", "falso"],
"ácido base débil" : ["débil", "grado", "equilibrio", "HA", "disocia", "monoprótico", "basicidad", "acidez"],
"pares conjugados" : "conjugad",
"anfóteros" : "anfote",
"bronsted-lowry": "lowry",
"pH resultante disolución" : "neutro",
"hidrólisis sales" : ["NaOH", "NH4Cl", "CH3COONa", "NaCl", "KCl",
"sales"],
"volumen agua a añadir" : "volumen de agua",
}
```

```python
def classify_exercise(statement, labels_dict):
    labels = []
    if pd.notna(statement):
        statement = statement.strip()
        for label, keywords in labels_dict.items():
            for kw in keywords:
                if kw in statement:
                    labels.append(label)
                    break
        if len(labels) > 0:
            return labels
    return None
```
